```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sentences = [
    'I love programming in Python',
    'Python is great for data science',
    'Deep learning is fascinating',
    'I enjoy solving problems with code',
    'Data analysis is crucial for insights'
]
```

```python
sentences
```

```python
## Define the vocabulary size
vocab_size = 1000 
```

```python
## One Hot Representation
one_hot_words = [one_hot(words,vocab_size) for words in sentences]
one_hot_words
```

```python
## word Embedding Representation
from tensorflow.keras.layers import Embedding
# from tensorflow.keras.processing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sentences_length = 8
embedded_docs = pad_sequences(one_hot_words, maxlen=sentences_length, padding='pre', )
print(embedded_docs)
```

```python
## feature representation
model = Sequential()
model.add(Embedding(vocab_size, 10, input_length=sentences_length))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
